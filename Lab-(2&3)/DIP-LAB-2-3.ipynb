# 4-Adjacency:

```python
  import cv2
def find_adjacent_pixels_4(img, x, y):
    height, width = img.shape[:2]
    adjacent_pixels = []
    if x > 0:
        adjacent_pixels.append((x - 1, y))  # Left
    if x < width - 1:
        adjacent_pixels.append((x + 1, y))  # Right
    if y > 0:
        adjacent_pixels.append((x, y - 1))  # Top
    if y < height - 1:
        adjacent_pixels.append((x, y + 1))  # Bottom
    return adjacent_pixels
# Example usage
image = cv2.imread('starryNight.jpg', cv2.IMREAD_GRAYSCALE)
adjacent_pixels = find_adjacent_pixels_4(image, x=5, y=5)
print(adjacent_pixels)      
```

# m-Adjacency:

```python
import cv2
def find_adjacent_pixels_8(img, x, y):
    height, width = img.shape[:2]
    adjacent_pixels = []
    for dx in range(-1, 2):
        for dy in range(-1, 2):
            if dx == 0 and dy == 0:
                continue
            nx, ny = x + dx, y + dy
            if 0 <= nx < width and 0 <= ny < height:
                adjacent_pixels.append((nx, ny))
    return adjacent_pixels
# Example usage
image = cv2.imread('starryNight.jpg', cv2.IMREAD_GRAYSCALE)
adjacent_pixels = find_adjacent_pixels_8(image, x=5, y=5)
print(adjacent_pixels)
```

# Convolution and corelation:

```python
import cv2
import numpy as np

# Load the image
image = cv2.imread('starryNight.jpg', cv2.IMREAD_GRAYSCALE)

# Define a kernel for convolution
kernel = np.array([[1, 1, 1],
                    [1, -8, 1],
                    [1, 1, 1]])

# Perform convolution
convolution_result = cv2.filter2D(image, -1, kernel)

# Perform correlation (template matching)
template = cv2.imread('starryNight.jpg', cv2.IMREAD_GRAYSCALE)
correlation_result = cv2.matchTemplate(image, template, cv2.TM_CCOEFF_NORMED)

# Display the results
cv2.imshow('Original Image', image)
cv2.imshow('Convolution Result', convolution_result)
cv2.imshow('Correlation Result', correlation_result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Averaging Filters:


```python
import cv2
import numpy as np

# Load the image
image = cv2.imread('starryNight.jpg')

# Define a 5x5 averaging kernel
kernel_size = 5
kernel = np.ones((kernel_size, kernel_size), np.float32) / (kernel_size * kernel_size)

# Apply the kernel using convolution
filtered_image = cv2.filter2D(image, -1, kernel)

# Display the original and filtered images
cv2.imshow('Original Image', image)
cv2.imshow('Filtered Image', filtered_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Mean Filter:

```python
import cv2
import numpy as np

# Load the image
image = cv2.imread('starryNight.jpg')

# Define a kernel size
kernel_size = 5

# Apply the mean filter
filtered_image = cv2.blur(image, (kernel_size, kernel_size))

# Display the original and filtered images
cv2.imshow('Original Image', image)
cv2.imshow('Mean Filtered Image', filtered_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Median Filter:

```python
import cv2
import numpy as np

# Load the image
image = cv2.imread('starryNight.jpg')

# Define a kernel size
kernel_size = 5

# Apply the median filter
filtered_image = cv2.medianBlur(image, kernel_size)

# Display the original and filtered images
cv2.imshow('Original Image', image)
cv2.imshow('Median Filtered Image', filtered_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
